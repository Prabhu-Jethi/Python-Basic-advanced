1) Wap to check string palindrome

```python
s = input("Enter a string: ")

reverse = s[::-1]

if s == reverse:
    print("Palindrome")
else:
    print("Not palindrome")
```

2) Print all unique elements in array

```python
arr = [2, 2, 4, 2, 4, 65, 18]
unique = []
seen = set()

for i in arr:
    if i not in seen:
        seen.add(i)
        unique.append(i)

print(unique)
```

6) To check whether the given year is Leap year or not

```python
year = int(input("Enter a year: "))

if (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0):
    print("Leap year")
else:
    print("Not a leap year")
```

8) To check whether the given number is prime or not

```python
n = int(input("Enter a number: "))

for i in range(2, int(n ** 0.5) + 1):
    if n % i == 0:
        print("Not Prime")
        break
    else:
        print("Prime")
        break
```

9) To check whether given number is strong number or not.

```python
n = int(input("Enter a number: "))
temp = n   # orginal no stored as temp
sum = 0    # sum of factorials of digit 

for i in str(n):  # convert num to str
    fact = 1      # start factorial
    for i in range(1, int(i)+1):  # calculates factorial of each digit
        fact *= i
    sum += fact     # add factorial to sum

if sum == temp:    # checks sum of fact == orginal no
    print("Strong number")
else:
    print("Not a strong number")

# del str
```

10) GCD of 2 numbers

```python
a = int(input("Enter 1st number: "))
b = int(input("Enter 2nd number: "))
gcd = 1

for i in range(1, min(a, b) + 1):
    if a % i == 0 and b % i == 0:
        gcd = i
print(gcd)
```

11) LCM of 2 numbers

```python
a = int(input("Enter 1st number: "))
b = int(input("Enter 2nd number: "))

x, y = a, b

while y != 0:
    x, y = y, x % y

gcd = x
lcm = (a * b) // gcd
print(lcm)

   # or 

# import math

# a = int(input("Enter 1st number: "))
# b = int(input("Enter 2nd number: "))

# print("LCM =", abs(a*b) // math.gcd(a,b))
```

14) Check Armstrong number

```python
n = (input("Enter a number: "))
a = 0
b = len((n))

i = 0
while i < len((n)):
    a = a + int(n[i]) ** b
    i = i + 1
if a == int(n):
    print("Armstrong")
else:
    print("Not Armstrong")
```
